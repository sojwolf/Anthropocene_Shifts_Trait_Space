## Longevity Data

```python
# read ~/Func_Neo_Phytes/Data/TRY/47270_17022026112317/47270.txt as pd
import pandas as pd

df = pd.read_csv("~/Func_Neo_Phytes/Data/TRY/47270_17022026112317/47270.txt", sep="\t", encoding='latin1', low_memory=False)
df.head()
```

```python
# number of unique values in "AccSpeciesName"
df['AccSpeciesName'].nunique()
```

```python
# column names
df.columns
```

```python
# filter for columns 'StdValueStr', 'OrigValueStr', 'OrigUnitStr', 'TraitID', 'AccSpeciesID', 'AccSpeciesName'
df_longevity = df[['StdValueStr', 'OrigValueStr', 'OrigUnitStr', 'TraitID', 'AccSpeciesID', 'AccSpeciesName', 'Comment']]
df_longevity.head(100)
```

```python
# filter for "annual" or "perennial" in "OrigValueStr"
df_longevity_cat = df_longevity[df_longevity['OrigValueStr'].str.strip().str.casefold().isin(['annual', 'perennial'])]
df_longevity_cat.head()
```

```python
df_longevity_cat.shape
```

```python
# keep only columns "AccSpeciesName" and "OrigValueStr"
df_longevity_cat = df_longevity_cat[['AccSpeciesName', 'OrigValueStr']]
df_longevity_cat.head()
```

```python
# filter for unique rows
df_longevity_cat = df_longevity_cat.drop_duplicates()
df_longevity_cat.head()
```

```python
# get dimensions of df_longevity_cat
df_longevity_cat.shape
```

```python
# number of unique values for "AccSpeciesName"
df_longevity_cat['AccSpeciesName'].nunique()
```

```python
# save df_longevity_cat as csv
df_longevity_cat.to_csv("~/Func_Neo_Phytes/Data/TRY/47270_17022026112317/47270_longevity_categorical.csv", index=False)
```
