### Add biome

```r
# read /net/scratch/swolf/GBIF/gbif_res3_try6_CS_wcvp_summary_stats_inkl_pca_mean_group_region_log10.csv
regions <- fread("/net/scratch/swolf/GBIF/gbif_res3_try6_CS_wcvp_summary_stats_inkl_pca_mean_group_region_log10.csv")
```

```r
head(regions)
```

```r
# filter regions for columns "region", "h3_res3"
regions <- regions[, .(region, h3_res3)]
```

```r
# remove duplicate rows in regions
regions <- unique(regions)
dim(regions)
```

```r
# add "region" column from regions with gbif_try_glonafon "h3_res3" column -left join
gbif_try_glonaf_region <- merge(gbif_try_glonaf, regions[, .(h3_res3, region)], by.x = "h3_res3", by.y = "h3_res3", all.x = TRUE)
head(gbif_try_glonaf_region)
```

```r
# split "region" column into realm and biome with "_" as separator
gbif_try_glonaf_region[, c("biome", "realm") := tstrsplit(region, "_", fixed = TRUE)]
head(gbif_try_glonaf_region)
```

## Most common introduced species

```r
# get most common speciesKey with counts for native and non-native statuses
most_common_species <- gbif_try_glonaf_region[, .(
  total_count = .N,
  native_count = sum(majority_status == "native", na.rm = TRUE),
  non_native_count = sum(majority_status == "non-native", na.rm = TRUE)
), by = .(speciesKey, realm)][order(-non_native_count)]
head(most_common_species, 10)
```

```r
# get top 10 most common non-native species for each realm
top_non_native_species_by_realm <- gbif_try_glonaf_region[majority_status == "non-native", .N, by = .(speciesKey, realm)][order(-N), head(.SD, 10), by = realm]
head(top_non_native_species_by_realm, 20)
```

```r
# save top_non_native_species_by_realm to csv
fwrite(top_non_native_species_by_realm, "/net/scratch/swolf/GBIF/top_non_native_species_by_realm.csv")
```
