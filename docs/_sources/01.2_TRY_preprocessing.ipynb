## Calculate species trait means

```r
library(dplyr)

# Group the data by Species and calculate the mean for each trait
TRY_species_means <- TRY %>%
  group_by(speciesKey) %>%  # Use Species Key !!
  summarise(across(starts_with("X"), ~ mean(.x, na.rm = TRUE))) # Use `across()` to get the mean of all trait columns
```

```r
head(TRY_species_means)
```

```r
total_missingness <- sum(is.na(TRY_species_means)) / prod(dim(TRY_species_means))
print(total_missingness)
```

```r
# Calculate missingness per column
column_missingness <- round(colSums(is.na(TRY_species_means)) / nrow(TRY_species_means), 2)
print(column_missingness)
mean(column_missingness)
```

```r
fwrite(TRY_species_means, "../../Results/Data/01_try6_gapfilled_species_means.csv", row.names = F, quote = T)
```
